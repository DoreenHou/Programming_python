 ### Q1 資料list變成一個Series  
 使用pandas把下方的資料list變成一個Series  
 將Series輸出到命令行  
 courses = ["chinese", "Math", "English", "Computer"]

```python
import pandas as pd

courses = ["chinese", "Math", "English", "Computer"]

data = pd.Series(data =courses)

print(data)
```

### Q4 將Series轉換成dataframe  
input: 學習成績Series   
  
chinese   80  
Math      90  
English   85  
Computer 100  
  
output: dataframe  
        grades   
chinese   80  
Math      90  
English   85  
Computer 100  
使用pandas把下方的資料Series變成一個dataframe  
資料列命名為grade,將List輸出到命令行  

```python
import pandas as pd

grades = {"chinese":80, "Math":90, "English":85, "Computer":100}
data = pd.Series(data = grades)

#start here
df = pd.DataFrame(data, columns=['grades'])
print(df)
```

### Q5 用Numpy建立Series    
類型float  
數值10-90,每兩個數字間隔10  

```python
import numpy as np
import pandas as pd

s = pd.Series(
    np.arange(10, 100, 10),#數值： 10-90 間隔10
    index=np.arange(101, 110), #索引： 101-109 間隔1
    dtype='float'#類型float
    )
print(s)
```

### Q6 轉換Series的資料類型  

```python
import pandas as pd
s = pd.Series(
    data=["001", "002", "003", "004"],
    index=list("abcd")
)
print(s)
s = s.astype(int) #轉換類型
print(s)
```

```python
s = s.map(int) #轉換類型
print(s)
```

### Q7 Sreies添加元素  

```python
import pandas as pd 

grades = {"chinese":80, "Math":90, "English":85, "Computer":100}
data = pd.Series(data = grades)

#start here
new_grades = pd.Series({
    "Physical": 88,
    "Chemistry": 95
})
data = pd.concat([data, new_grades])
print(data)
```

### Q8 Series轉換成dataframe  

```python
import pandas as pd

grades = {"chinese":80, "Math":90, "English":85, "Computer":100}
data = pd.Series(data = grades)

#start here
df = data.reset_index() #重置索引
df.columns = ["course", "grade"] #.columns 給予索引標籤

print(df)
```

### Q9 建立一個DataFrame
