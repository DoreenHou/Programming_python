```python
# dropna 用於删除包含缺失值（NaN或None）的行或列的方法
df = df.dropna(subset=["name"])
df
```

# missing values

```python
# list of NaN values by columns
# isna() 用於檢查DataFrame或Series中的每个元素是否为缺失值（NaN）
df.isna().sum()
```

```python
# age NaN values
```

```python
pd.pivot_table(df,
               index = "pclass",
               values = "age")
```

```python
df
```

```python
# replace the NaN age by the mean columns by the pclass categoty
df['age'] = df['age'].fillna(df.groupby('pclass')['age'].transform('mean')) 
df
```

```python
df.columns
```

```python
# drop columns if number of columns NaN is > 10% of observation
#list of columns
cols = df.columns.tolist()
for col in cols:
    print(col)
    x = df[col].isna().sum()
    p = x/df.shape[0]
    if p > 0.1 :
        df = df.drop(col, axis=1)

df
```

```python
# drop columns if number of columns NaN is > 10% of observation
#list of columns

cols = df.columns.tolist()
for col in cols:
    p = df[col].isna().sum()/df.shape[0]
    if p > 0.1 :
        df = df.drop(col, axis=1)

df.isna().sum()
```

```python
# replace the NaN fare by the mean columns by the pclass categoty
df['fare'] = df['fare'].fillna(df.groupby('pclass')['fare'].transform('mean')) 
```

```python
df["embarked"] = df["embarked"].fillna(df["embarked"].mode()[0])
```

```python
df.isna().sum()
```

```python
# create new columns "age_cat"
# if age < 18 = minor else Adulit
df["age_cat"] = df['age'].apply(lambda x: 'adult' if x > 18 else 'minor')
```

```python
#會考
#create new columns "age_class"
df['age_class'] = pd.cut(df['age'],
                          bins=[0, 20, 40, 60, 100],
                          labels=['Class D', 'Class C', 'Class B', 'Class A'])
```
